## Treinamento Modelo


```python
import numpy as np
from sklearn.model_selection import train_test_split


cat_cols = df.select_dtypes(include=["category"]).columns
for col in cat_cols:
    df[col] = df[col].astype(str).astype("category")

df['renda_V7616'] = df['renda_V7616'].replace(0, 0.001)

X = df.drop(columns=['renda_V7616', 'renda_sal_V7617'])
Y = df['renda_V7616']

X_train, X_test, y_train, y_test = train_test_split(
    X, Y, test_size=0.33, random_state=42)
```

```python
df.info()
```

```python
Y.info()
```

```python
import gc
import optuna
from sklearn.model_selection import cross_val_score
from xgboost import XGBRegressor

def objective(trial):
    reg = XGBRegressor(
        enable_categorical=True,
        tree_method="hist",
        random_state=42,
        n_estimators=trial.suggest_int("n_estimators", 100, 1000),
        learning_rate=trial.suggest_float("learning_rate", 0.01, 0.2, log=True),
        max_depth=trial.suggest_int("max_depth", 3, 7),
        subsample=trial.suggest_float("subsample", 0.6, 1.0),
        objective="reg:absoluteerror",
        max_cat_threshold=32,
    )
    score = cross_val_score(reg, X_train, y_train, cv=3, scoring="d2_absolute_error_score")
    resultado = score.mean()  # sem o sinal de menos

    del reg, score
    gc.collect()

    return resultado

study = optuna.create_study(direction="maximize")  # maximizar o D²
study.optimize(objective, n_trials=20, gc_after_trial=True)
```

```python
print(study.best_trial.value, study.best_trial.params)
```
